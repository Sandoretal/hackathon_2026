The dataset variables:

1. PATNO: patient's number
2. SEX: 1=Male, 0=Female 
3. HANDED: Handedness: 1=Right, 2=Left, 3=Mixed
4. COHORT_DEFINITION: "Parkinson's Disease" / "Prodromal"
5. ENROLL_AGE: patient's age at enrolment (baseline)
6. NP3TOT: patient's UPDRS-III score 
7. MCATOT: patient's MoCA score 
8. GMvol: patient's total grey matter volume (mm^3) for patients with an MRI scan at baseline, for the rest this column is NaN.
9. clinic_id: ID of the clinic the patient is enrolled in (regions are in data/clinics_lookup_dirty.csv)


Severity Scales:

UPDRS-III scale: Range - 0-132,   Mild 0-32;   Moderate 33-58; Severe 59+

MoCA scale:     Range - 0-30, Normal 26-30;   Mild 18-25; Moderate 10-17; Severe <10
